```python
# ============================================================
# CONIC SECTION IDENTIFIER (D7)
# Cell 1: Imports
# ============================================================

import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

from math import sqrt, atan2, degrees, isclose

np.set_printoptions(precision=5, suppress=True)
```

```python
# ============================================================
# Cell 2: INPUT CONFIGURATION
# ============================================================

# General conic:
#
# ax^2 + 2hxy + by^2 + 2gx + 2fy + c = 0

a = 1
h = 0
b = 1
g = 0
f = 0
c = -9

# Numerical tolerance
TOL = 1e-9

print("Input equation:")
print(f"{a}x² + {2*h}xy + {b}y² + {2*g}x + {2*f}y + {c} = 0")
```

```python
# ============================================================
# Cell 3: MATHEMATICAL MODEL
# ============================================================

# Quadratic matrix A
A = np.array([
    [a, h],
    [h, b]
], dtype=float)

# Linear coefficient vector d
d = np.array([g, f], dtype=float)

# Constant
constant = float(c)

print("Quadratic matrix A:")
print(A)

print("\nLinear vector d:")
print(d)

print("\nOriginal conic:")
print("xᵀ A x + 2 dᵀ x + c = 0")
```

```python
# ============================================================
# Cell 4: EIGENVALUE / EIGENVECTOR CALCULATION
# ============================================================

# A is symmetric, so eigh() gives real eigenvalues
# and orthonormal eigenvectors.
eigenvalues, Q = np.linalg.eigh(A)

print("Eigenvalues:")
print(eigenvalues)

print("\nOrthogonal transformation matrix Q:")
print(Q)

print("\nCheck QᵀQ:")
print(Q.T @ Q)

print("\nDiagonal form Qᵀ A Q:")
print(Q.T @ A @ Q)
```

```python
# ============================================================
# Cell 5: MATHEMATICAL ENGINE
# ============================================================

def analyze_conic(a, h, b, g, f, c, tol=1e-9):

    A = np.array([[a, h],
                  [h, b]], dtype=float)

    d = np.array([g, f], dtype=float)

    # Eigenvalue decomposition
    eigenvalues, Q = np.linalg.eigh(A)

    # Sort eigenvalues by magnitude for easier interpretation
    order = np.argsort(np.abs(eigenvalues))
    eigenvalues = eigenvalues[order]
    Q = Q[:, order]

    determinant = np.linalg.det(A)
    matrix_rank = np.linalg.matrix_rank(A, tol)

    result = {
        "A": A,
        "d": d,
        "eigenvalues": eigenvalues,
        "Q": Q,
        "det_A": determinant,
        "rank_A": matrix_rank,
        "classification": None,
        "center": None,
        "vertex": None,
        "axes": None,
        "rotation_angle": None,
        "parameter": None,
        "F": None,
        "canonical_equation": None
    }

    # --------------------------------------------------------
    # CASE 1: A is full rank
    # --------------------------------------------------------

    if matrix_rank == 2:

        # Center:
        # A x0 + d = 0
        center = -np.linalg.solve(A, d)

        # F = c - dᵀ A⁻¹ d
        F = c - d @ np.linalg.solve(A, d)

        result["center"] = center
        result["F"] = F

        # Circle test
        is_circle = (
            abs(a - b) < tol and
            abs(h) < tol and
            abs(a) > tol
        )

        # ----------------------------------------------------
        # Positive/negative definite A
        # ----------------------------------------------------

        if determinant > tol:

            # λ1 and λ2 have the same sign
            transformed_constants = -F / eigenvalues

            if abs(F) <= tol:

                # A definite and F = 0
                # Only the center satisfies the equation.
                result["classification"] = "Point"

                result["canonical_equation"] = (
                    "λ₁u² + λ₂v² = 0"
                )

            elif np.all(transformed_constants > tol):

                if is_circle:
                    result["classification"] = "Circle"
                else:
                    result["classification"] = "Ellipse"

                semi_axes = np.sqrt(transformed_constants)

                # semi_axes correspond to eigenvalues
                result["axes"] = semi_axes

                # Major axis = largest semi-axis
                major_index = np.argmax(semi_axes)

                eigenvector = Q[:, major_index]

                angle = degrees(
                    atan2(eigenvector[1], eigenvector[0])
                )

                result["rotation_angle"] = angle

                result["canonical_equation"] = (
                    f"u²/{semi_axes[0]**2:.5f} + "
                    f"v²/{semi_axes[1]**2:.5f} = 1"
                )

            else:

                result["classification"] = "Empty"

                result["canonical_equation"] = (
                    "No real points"
                )

        # ----------------------------------------------------
        # Indefinite A
        # ----------------------------------------------------

        elif determinant < -tol:

            if abs(F) <= tol:

                # Degenerate hyperbola:
                # λ1u² + λ2v² = 0
                result["classification"] = "Pair of Lines"

                result["canonical_equation"] = (
                    "λ₁u² + λ₂v² = 0"
                )

            else:

                result["classification"] = "Hyperbola"

                transformed_constants = -F / eigenvalues

                semi_axes = np.sqrt(np.abs(transformed_constants))

                result["axes"] = semi_axes

                # Transverse axis corresponds to positive
                # squared coordinate
                positive_indices = np.where(
                    transformed_constants > 0
                )[0]

                if len(positive_indices) > 0:

                    transverse_index = positive_indices[0]

                    eigenvector = Q[:, transverse_index]

                    angle = degrees(
                        atan2(
                            eigenvector[1],
                            eigenvector[0]
                        )
                    )

                    result["rotation_angle"] = angle

                result["canonical_equation"] = (
                    f"u²/{semi_axes[0]**2:.5f} "
                    f"+ v²/{semi_axes[1]**2:.5f} = 1"
                )

    # --------------------------------------------------------
    # CASE 2: A has rank 1
    # --------------------------------------------------------

    elif matrix_rank == 1:

        # One eigenvalue is approximately zero.
        nonzero_index = np.argmax(np.abs(eigenvalues))
        zero_index = 1 - nonzero_index

        lam = eigenvalues[nonzero_index]

        # Transform linear vector
        transformed_d = Q.T @ d

        d_u = transformed_d[nonzero_index]
        d_v = transformed_d[zero_index]

        # ----------------------------------------------------
        # Parabola
        # ----------------------------------------------------

        if abs(d_v) > tol:

            result["classification"] = "Parabola"

            # Vertex in transformed coordinates
            u_vertex = -d_u / lam

            v_vertex = -(
                c - (d_u ** 2) / lam
            ) / (2 * d_v)

            uv_vertex = np.zeros(2)

            uv_vertex[nonzero_index] = u_vertex
            uv_vertex[zero_index] = v_vertex

            # Convert vertex back to x,y coordinates
            vertex = Q @ uv_vertex

            result["vertex"] = vertex

            # In local coordinates:
            #
            # U² = 4pV
            #
            # p = -d_v / (2λ)

            p = -d_v / (2 * lam)

            result["parameter"] = p

            # Parabola axis is along null eigenvector
            axis_vector = Q[:, zero_index]

            angle = degrees(
                atan2(axis_vector[1], axis_vector[0])
            )

            result["rotation_angle"] = angle

            result["canonical_equation"] = (
                f"U² = {4*p:.5f}V"
            )

        # ----------------------------------------------------
        # Degenerate rank-1 cases
        # ----------------------------------------------------

        else:

            F = c - (d_u ** 2) / lam

            result["F"] = F

            value = -F / lam

            if value > tol:

                result["classification"] = "Parallel Lines"

                distance = sqrt(value)

                result["axes"] = np.array([distance])

                result["canonical_equation"] = (
                    f"(u - u₀)² = {value:.5f}"
                )

            elif abs(value) <= tol:

                result["classification"] = "Single Line"

                result["canonical_equation"] = (
                    "(u - u₀)² = 0"
                )

            else:

                result["classification"] = "Empty"

                result["canonical_equation"] = (
                    "No real points"
                )

    # --------------------------------------------------------
    # CASE 3: A is zero matrix
    # --------------------------------------------------------

    else:

        # Equation becomes:
        #
        # 2gx + 2fy + c = 0

        if np.linalg.norm(d) > tol:

            result["classification"] = "Line"

            result["canonical_equation"] = (
                "2gx + 2fy + c = 0"
            )

        elif abs(c) <= tol:

            result["classification"] = "Whole Plane"

            result["canonical_equation"] = (
                "0 = 0"
            )

        else:

            result["classification"] = "Empty"

            result["canonical_equation"] = (
                "c = 0 is impossible"
            )

    return result
```

```python
# ============================================================
# Cell 6: RUN ANALYSIS
# ============================================================

result = analyze_conic(
    a, h, b, g, f, c, TOL
)

print("Analysis completed.")
print("Classification:", result["classification"])
```

```python
# ============================================================
# Cell 7: OUTPUT DISPLAY
# ============================================================

def print_result(result):

    print("=" * 65)
    print("             CONIC SECTION IDENTIFIER")
    print("=" * 65)

    print("\n1. Quadratic Matrix A")
    print(result["A"])

    print("\n2. Eigenvalues")
    print(result["eigenvalues"])

    print("\n3. Eigenvectors / Orthogonal Matrix Q")
    print(result["Q"])

    print("\n4. Determinant of A")
    print(f"{result['det_A']:.8f}")

    print("\n5. Rank of A")
    print(result["rank_A"])

    print("\n6. Classification")
    print(">>>", result["classification"])

    if result["center"] is not None:
        print("\n7. Center")
        print(
            f"({result['center'][0]:.5f}, "
            f"{result['center'][1]:.5f})"
        )

    if result["vertex"] is not None:
        print("\n7. Vertex")
        print(
            f"({result['vertex'][0]:.5f}, "
            f"{result['vertex'][1]:.5f})"
        )

    if result["axes"] is not None:

        print("\n8. Axis information")

        for i, value in enumerate(result["axes"]):
            print(
                f"Axis {i+1}: {value:.5f}"
            )

    if result["rotation_angle"] is not None:
        print("\n9. Rotation angle")
        print(
            f"{result['rotation_angle']:.5f} degrees"
        )

    print("\n10. Canonical form")
    print(result["canonical_equation"])

    print("=" * 65)


print_result(result)
```

```python
# ============================================================
# Cell 8: MATHEMATICAL STEPS
# ============================================================

print("STEP 1: Original equation")
print("ax² + 2hxy + by² + 2gx + 2fy + c = 0")

print("\nSTEP 2: Quadratic form")
print("xᵀ A x + 2dᵀx + c = 0")

print("\nA =")
print(A)

print("\nd =")
print(d)

print("\nSTEP 3: Eigenvalue decomposition")
print("A = Q Λ Qᵀ")

print("\nEigenvalues:")
print(eigenvalues)

print("\nQ =")
print(Q)

print("\nSTEP 4: Orthogonal transformation")
print("x = Q u")

print("\nSTEP 5: Rotated canonical system")
print("Qᵀ A Q = Λ")

print("\nResult:")
print(Q.T @ A @ Q)

print("\nSTEP 6: Final classification")
print(result["classification"])
```

```python
# ============================================================
# Cell 9: CONIC PLOT
# ============================================================

def plot_conic(a, h, b, g, f, c, result):

    classification = result["classification"]

    # --------------------------------------------------------
    # Choose plotting center
    # --------------------------------------------------------

    if result["center"] is not None:
        plot_center = result["center"]

    elif result["vertex"] is not None:
        plot_center = result["vertex"]

    else:
        plot_center = np.array([0.0, 0.0])

    # --------------------------------------------------------
    # Choose plot size
    # --------------------------------------------------------

    if result["axes"] is not None:
        scale = max(np.max(np.abs(result["axes"])), 1.0)

    elif result["parameter"] is not None:
        scale = max(abs(result["parameter"]) * 5, 3.0)

    else:
        scale = 5.0

    # Hyperbolas need a wider viewing region
    if classification == "Hyperbola":
        scale = max(scale * 4, 6)

    elif classification == "Parabola":
        scale = max(scale * 2, 5)

    else:
        scale = max(scale * 1.5, 4)

    x_min = plot_center[0] - scale
    x_max = plot_center[0] + scale

    y_min = plot_center[1] - scale
    y_max = plot_center[1] + scale

    x = np.linspace(x_min, x_max, 600)
    y = np.linspace(y_min, y_max, 600)

    X, Y = np.meshgrid(x, y)

    # Evaluate original conic equation
    Z = (
        a * X**2
        + 2*h*X*Y
        + b * Y**2
        + 2*g*X
        + 2*f*Y
        + c
    )

    plt.figure(figsize=(9, 7))

    # Plot implicit curve
    contour = plt.contour(
        X, Y, Z,
        levels=[0],
        linewidths=2
    )

    # Coordinate axes
    plt.axhline(0, linewidth=0.8)
    plt.axvline(0, linewidth=0.8)

    # Mark center
    if result["center"] is not None:

        plt.scatter(
            result["center"][0],
            result["center"][1],
            s=70,
            marker="o",
            label="Center"
        )

    # Mark vertex
    if result["vertex"] is not None:

        plt.scatter(
            result["vertex"][0],
            result["vertex"][1],
            s=70,
            marker="o",
            label="Vertex"
        )

    plt.xlabel("x")
    plt.ylabel("y")

    plt.title(
        f"Conic Section: {classification}"
    )

    plt.grid(True, alpha=0.25)
    plt.axis("equal")
    plt.legend()

    plt.show()


plot_conic(
    a, h, b, g, f, c, result
)
```

```python
# ============================================================
# Cell 10: EIGENVALUE CHART
# ============================================================

def plot_eigenvalues(result):

    eigenvalues = result["eigenvalues"]

    labels = [
        "λ₁",
        "λ₂"
    ]

    plt.figure(figsize=(7, 5))

    plt.bar(
        labels,
        eigenvalues
    )

    plt.axhline(
        0,
        linewidth=1
    )

    plt.ylabel("Eigenvalue")
    plt.title(
        "Eigenvalues of the Quadratic Matrix"
    )

    plt.grid(
        axis="y",
        alpha=0.25
    )

    plt.show()


plot_eigenvalues(result)
```

```python
# ============================================================
# Cell 11: EIGENVECTOR / ROTATION VISUALIZATION
# ============================================================

def plot_eigenvectors(result):

    Q = result["Q"]

    plt.figure(figsize=(7, 7))

    plt.axhline(0, linewidth=0.8)
    plt.axvline(0, linewidth=0.8)

    # Eigenvector 1
    plt.quiver(
        0, 0,
        Q[0, 0],
        Q[1, 0],
        angles="xy",
        scale_units="xy",
        scale=1,
        width=0.008,
        label="Eigenvector 1"
    )

    # Eigenvector 2
    plt.quiver(
        0, 0,
        Q[0, 1],
        Q[1, 1],
        angles="xy",
        scale_units="xy",
        scale=1,
        width=0.008,
        label="Eigenvector 2"
    )

    plt.xlim(-1.3, 1.3)
    plt.ylim(-1.3, 1.3)

    plt.xlabel("x")
    plt.ylabel("y")

    plt.title(
        "Orthogonal Eigenvector Directions"
    )

    plt.grid(True, alpha=0.25)
    plt.axis("equal")
    plt.legend()

    plt.show()


plot_eigenvectors(result)
```

```python
# ============================================================
# CELL 18: IMPROVED PROFESSIONAL CONIC GRAPH
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

def professional_conic_plot(
    a, h, b, g, f, c,
    result,
    grid_size=700
):
    """
    Professional visualization of the conic section.

    Equation:
        ax² + 2hxy + by² + 2gx + 2fy + c = 0

    Uses the ORIGINAL equation for contour plotting.
    """

    classification = result["classification"]

    # --------------------------------------------------------
    # 1. Select important geometric point
    # --------------------------------------------------------

    if result.get("center") is not None:
        reference_point = np.asarray(
            result["center"], dtype=float
        )

    elif result.get("vertex") is not None:
        reference_point = np.asarray(
            result["vertex"], dtype=float
        )

    else:
        reference_point = np.array([0.0, 0.0])

    # --------------------------------------------------------
    # 2. Determine plotting scale automatically
    # --------------------------------------------------------

    if result.get("axes") is not None:
        axes = np.asarray(result["axes"], dtype=float)

        finite_axes = axes[
            np.isfinite(axes)
        ]

        if len(finite_axes) > 0:
            scale = max(
                4.0,
                float(np.max(np.abs(finite_axes))) * 2.5
            )
        else:
            scale = 6.0

    elif result.get("parameter") is not None:

        scale = max(
            6.0,
            abs(float(result["parameter"])) * 5
        )

    else:
        scale = 6.0

    # Wider view for hyperbola
    if classification == "Hyperbola":
        scale = max(scale, 10.0)

    # Wider view for parabola
    if classification == "Parabola":
        scale = max(scale, 8.0)

    # --------------------------------------------------------
    # 3. Create coordinate grid
    # --------------------------------------------------------

    x_min = reference_point[0] - scale
    x_max = reference_point[0] + scale

    y_min = reference_point[1] - scale
    y_max = reference_point[1] + scale

    x = np.linspace(
        x_min,
        x_max,
        grid_size
    )

    y = np.linspace(
        y_min,
        y_max,
        grid_size
    )

    X, Y = np.meshgrid(x, y)

    # --------------------------------------------------------
    # 4. Original conic equation
    # --------------------------------------------------------

    Z = (
        a * X**2
        + 2 * h * X * Y
        + b * Y**2
        + 2 * g * X
        + 2 * f * Y
        + c
    )

    # --------------------------------------------------------
    # 5. Create professional figure
    # --------------------------------------------------------

    fig, ax = plt.subplots(
        figsize=(10, 8)
    )

    # --------------------------------------------------------
    # 6. Plot the conic
    # --------------------------------------------------------

    contour = ax.contour(
        X,
        Y,
        Z,
        levels=[0],
        linewidths=3
    )

    # --------------------------------------------------------
    # 7. Coordinate axes
    # --------------------------------------------------------

    ax.axhline(
        0,
        linewidth=1.2,
        alpha=0.7
    )

    ax.axvline(
        0,
        linewidth=1.2,
        alpha=0.7
    )

    # --------------------------------------------------------
    # 8. Plot center
    # --------------------------------------------------------

    if result.get("center") is not None:

        center = np.asarray(
            result["center"],
            dtype=float
        )

        ax.scatter(
            center[0],
            center[1],
            s=100,
            marker="o",
            zorder=5,
            label=(
                f"Center "
                f"({center[0]:.2f}, "
                f"{center[1]:.2f})"
            )
        )

    # --------------------------------------------------------
    # 9. Plot vertex
    # --------------------------------------------------------

    if result.get("vertex") is not None:

        vertex = np.asarray(
            result["vertex"],
            dtype=float
        )

        ax.scatter(
            vertex[0],
            vertex[1],
            s=110,
            marker="*",
            zorder=6,
            label=(
                f"Vertex "
                f"({vertex[0]:.2f}, "
                f"{vertex[1]:.2f})"
            )
        )

    # --------------------------------------------------------
    # 10. Plot principal directions
    # --------------------------------------------------------

    Q = result.get("Q")

    if Q is not None and reference_point is not None:

        # Only display eigenvector directions for
        # non-degenerate conics.
        if classification in [
            "Ellipse",
            "Circle",
            "Hyperbola",
            "Parabola"
        ]:

            vector_length = scale * 0.65

            for i in range(2):

                direction = Q[:, i]

                start = (
                    reference_point
                    - vector_length * direction
                )

                end = (
                    reference_point
                    + vector_length * direction
                )

                ax.plot(
                    [start[0], end[0]],
                    [start[1], end[1]],
                    linestyle="--",
                    linewidth=1.2,
                    alpha=0.55,
                    label=(
                        f"Principal axis {i+1}"
                    )
                )

    # --------------------------------------------------------
    # 11. Labels
    # --------------------------------------------------------

    ax.set_xlabel(
        "x",
        fontsize=13
    )

    ax.set_ylabel(
        "y",
        fontsize=13
    )

    ax.set_title(
        "CONIC SECTION IDENTIFIER",
        fontsize=18,
        fontweight="bold",
        pad=15
    )

    # --------------------------------------------------------
    # 12. Subtitle containing classification
    # --------------------------------------------------------

    equation_text = (
        f"{a}x² + {2*h}xy + {b}y² "
        f"+ {2*g}x + {2*f}y + {c} = 0"
    )

    ax.text(
        0.5,
        1.015,
        f"Classification: {classification}",
        transform=ax.transAxes,
        ha="center",
        fontsize=14,
        fontweight="bold"
    )

    ax.text(
        0.5,
        -0.10,
        equation_text,
        transform=ax.transAxes,
        ha="center",
        fontsize=11
    )

    # --------------------------------------------------------
    # 13. Grid and appearance
    # --------------------------------------------------------

    ax.grid(
        True,
        linestyle=":",
        linewidth=0.8,
        alpha=0.45
    )

    ax.set_aspect(
        "equal",
        adjustable="box"
    )

    # --------------------------------------------------------
    # 14. Add legend only when necessary
    # --------------------------------------------------------

    handles, labels = ax.get_legend_handles_labels()

    if handles:
        ax.legend(
            loc="best",
            frameon=True,
            fontsize=9
        )

    # --------------------------------------------------------
    # 15. Information box
    # --------------------------------------------------------

    eigenvalues = result.get(
        "eigenvalues",
        None
    )

    if eigenvalues is not None:

        eigen_text = (
            f"λ₁ = {eigenvalues[0]:.4f}\n"
            f"λ₂ = {eigenvalues[1]:.4f}"
        )

        ax.text(
            0.02,
            0.97,
            eigen_text,
            transform=ax.transAxes,
            verticalalignment="top",
            fontsize=10,
            bbox=dict(
                boxstyle="round,pad=0.5",
                alpha=0.85
            )
        )

    # --------------------------------------------------------
    # 16. Final layout
    # --------------------------------------------------------

    plt.tight_layout()

    plt.show()
```
